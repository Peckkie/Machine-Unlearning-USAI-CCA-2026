```python
dataset = pd.read_csv("/home/kannika/code/mini-ImageNet_MachineUnlearn.csv", dtype=str)
print(dataset.shape)
dataset.head()
```

```python
'''
new path = /media/tohn/HDD2/miniImageNet_Dataset
'''

newpath = "/media/tohn/HDD2/miniImageNet_Dataset/"
oldpath = "/media/HDD/mini-ImageNet/"

dataset['img_path'] = dataset['img_path'].str.replace(oldpath, newpath)
print(dataset.shape)
dataset.head()
```

```python
from PIL import Image, ImageFile
from efficientnet.keras import center_crop_and_resize, preprocess_input
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
height=width = 224
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
      rescale=1./255,
      rotation_range=30,
      width_shift_range=0.2,
      height_shift_range=0.2,
      brightness_range=[0.5, 1.5],
      shear_range=0.4,
      zoom_range=0.2,
      horizontal_flip=False,
      fill_mode='nearest')

train_generator = train_datagen.flow_from_dataframe(
    dataframe = dataset,
    directory = None,
    x_col = 'img_path',
    y_col = 'cls',
    target_size = (height, width),
    batch_size=BATCH_SIZE,
    color_mode= 'rgb',
    class_mode='categorical')
```
