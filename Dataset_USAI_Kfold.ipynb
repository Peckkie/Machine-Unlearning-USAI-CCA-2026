```python
import numpy as np
import pandas as pd
```

## DATASET: Train Set

```python
dataframe = pd.read_csv('/media/tohn/HDD/VISION_dataset/Traindf_fold4_8_v1.csv')
print(dataframe.shape)
dataframe.head()
```

```python
print(set(dataframe.fold))
```

## DATASET: Validation Set

```python
dataval = pd.read_csv('/media/tohn/HDD/VISION_dataset/Valdf_fold3_v1.csv')
print(set(dataval.fold))
print(dataval.shape)
dataval.head()
```

## Combine DATASET: Train Set + Validation Set (Fold3-Fold10)

```python
# รวมข้อมูล (ต่อแถว)
df_merged = pd.concat([dataframe, dataval], ignore_index=True)
# ตรวจสอบขนาดของข้อมูลที่รวมแล้ว
print("Merged shape:", df_merged.shape)
print("Fold Number: ", set(df_merged.fold))
# แสดงข้อมูลตัวอย่าง
df_merged.head()
```

```python
df_merged.info()
```

```python
fold_df = df_merged[df_merged["fold"]==5].reset_index(drop=True)
print("Fold: ", set(fold_df.fold))
print(fold_df.shape)
fold_df.head()
```

```python
fold_df = df_merged[df_merged["fold"]!=5].reset_index(drop=True)
print("Fold: ", set(fold_df.fold))
print(fold_df.shape)
fold_df.head()
```

```python
df_merged.to_csv('/media/tohn/HDD/VISION_dataset/Traindf_fold3_10_Kfold.csv')
```
